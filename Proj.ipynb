# Split the data

```python
y_tech = df['Technology']
y_trans = df['Transportation']
X = df['Body']
```

```python
X = X.values.tolist()
y_trans = y_trans.values.tolist()
y_tech = y_tech.values.tolist()
```

```python
from sklearn.model_selection import train_test_split
import random 
random.seed(42)

#tech 
X_train_tech, X_test_tech, y_train_tech, y_test_tech = train_test_split(X, y_tech, test_size=.2, random_state=42)
#print(X_train.shape, X_test.shape)

#transportation
X_train_trans, X_test_trans, y_train_trans, y_test_trans = train_test_split(X, y_trans, test_size=.2, random_state=42)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import precision_score, recall_score, f1_score

#Technology

vec = CountVectorizer(ngram_range = (1,1))

X_train = vec.fit_transform(X_train_tech) 
X_test = vec.transform(X_test_tech)
```

# Implementing the features into the Model

```python
from collections import Counter

def count_capital_words(string):
    words = string.split()  # Split the string into words
    count = 0
    for word in words:
        if word.isupper():  # Check if the word is capitalized
            count += 1
    return count

def count_exclamation_points(string):
    count = 0
    for char in string:
        if char == '!':
            count += 1
    return count

def count_repeated_words(string):
    # Tokenize the string into words
    words = string.split()
    # Count occurrences of each word
    word_counts = Counter(words)
    # Count repeated words
    repeated_count = sum(count for count in word_counts.values() if count > 1)
    return repeated_count
```
